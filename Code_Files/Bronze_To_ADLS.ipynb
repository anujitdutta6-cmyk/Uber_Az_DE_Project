```python
import pandas as pd

files = [
    "map_cities",
    "map_cancellation_reasons",
    "bulk_rides",
    "map_payment_methods",
    "map_ride_statuses",
    "map_vehicle_makes",
    "map_vehicle_types"
]

for file in files:
    url = f"https://dluberprojectdev.blob.core.windows.net/raw/ingestion/{file}.json?"
   
    df = pd.read_json(url)
    df_spark = spark.createDataFrame(df)
    if not spark.catalog.tableExists("uber.bronze.bulk_rides"):
    # Writing Data Into Bronze Layer
        df_spark.write.format("delta")\
            .mode("overwrite")\
            .option("overwriteSchema", "true")\
            .saveAsTable(f"uber.bronze.bulk_rides")
        print("This will not run more than 1 time")
```
